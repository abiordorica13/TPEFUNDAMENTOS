```python
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
df = pd.read_excel("25-testcar-2026-01-21.xlsx")
df.head()#Asi vemos que el archivo se cargo correctamente
```

```python
df.info()
```

```python
#Para ver la cantidad de valores nulos 
nulos = pd.DataFrame({
    "Cantidad" : df.isnull().sum(),
    "Porcentaje" : df.isnull().mean()*100
})

nulos.sort_values("Porcentaje", ascending=False)
```

```python
df.describe()
```

```python
#para ver las filas duplicadas 
df.duplicated().sum()
```

```python
#Distribucion de la economia del combustible 

plt.figure(figsize=(8,5))
plt.hist(df["RND_ADJ_FE"].dropna(), bins=30, edgecolor='black')
plt.xlabel("MPG")
plt.ylabel("Frecuencia")
plt.title("Distribucion de la economia de combustible")

plt.show()
```

```python
#Distribucion de la potencia 

plt.figure(figsize=(8,5))
plt.hist(df["Rated Horsepower"].dropna(), bins=30, edgecolor='black')
plt.xlabel("Potencia (HP)")
plt.ylabel("Frecuencia")
plt.title("Distribución de la potencia")

plt.show()
```

```python
#Distribucion cilindrada 

plt.figure(figsize=(8,5))
plt.hist(df["Test Veh Displacement (L)"].dropna(), bins=30, edgecolor='black')
plt.xlabel("Cilindrada (L)")
plt.ylabel("Frecuencia")
plt.title("Distribución de la cilindrada")

plt.show()
```

```python
#Distribucion del peso

plt.figure(figsize=(8,5))
plt.hist(df["Equivalent Test Weight (lbs.)"].dropna(), bins=30, edgecolor='black')
plt.xlabel("Peso equivalente de prueba (lbs.)")
plt.ylabel("Frecuencia")
plt.title("Distribución del peso")

plt.show()
```

```python
#Distribucion de los sistemas de traccion

plt.figure(figsize=(8,5))

df["Drive System Description"].value_counts().plot(kind="barh", edgecolor="black") #value_counts() nos da la cantidad de cada sistema de traccion y plot(kind="barh") nos hace el grafico de barras

plt.xlabel("Frecuencia")
plt.ylabel("Sistema de tracción")
plt.title("Distribución de los sistemas de tracción")

plt.show()
```

```python
# Distribucion de los tipos de transmision

plt.figure(figsize=(8,5))

df["Tested Transmission Type"].value_counts().plot(kind="barh", edgecolor="black") #lo inverti para que se viera mejor

plt.xlabel("Frecuencia")
plt.ylabel("Transmisión")
plt.title("Distribución de los tipos de transmisión")
plt.show()
```

```python
#Distribucion del combustible

plt.figure(figsize=(8,5))

df["Test Fuel Type Description"].value_counts().plot(kind="barh", edgecolor="black") 

plt.xlabel("Frecuencia")
plt.ylabel("Combustible")
plt.title("Distribución de los tipos de combustible")

plt.show()
```

```python
# Cilindrada sin outliers centinela

col = 'Test Veh Displacement (L)'
disp_real = df[(df[col].notna()) & (df[col] < 10)][col]

plt.figure(figsize=(8, 5))
plt.hist(disp_real, bins=30, edgecolor="black")
plt.title('Distribución de la cilindrada (sin outliers centinela)')
plt.xlabel('Cilindrada (L)')
plt.ylabel('Frecuencia')
plt.show()
```

```python
# Economia del combustible sin outliers centinela

fe_real = df[(df["RND_ADJ_FE"].notna()) & (df["RND_ADJ_FE"] < 150)]["RND_ADJ_FE"]

plt.figure(figsize=(8, 5))
plt.hist(fe_real, bins=30, edgecolor='black')
plt.xlabel("MPG")
plt.ylabel("Frecuencia")
plt.title("Distribución de la economía de combustible (sin outliers centinela)")

plt.show()
```
