```python
import matplotlib.pyplot as plt

def graficar_histograma(df, columna, titulo, etiqueta_x, bins=30, limite=None):
    # Detectar si 'columna' es texto (nombre de la columna) o una variable con datos
    if isinstance(columna, str):
        datos = df[columna]
    else:
        datos = columna  # Toma directamente tu variable (ej. fe_real)
        
    # Aplicar límite y eliminar nulos
    if limite is not None:
        datos = datos[datos < limite]
        
    datos = datos.dropna()
        
    plt.figure(figsize=(8, 5))
    plt.hist(datos, bins=bins, edgecolor='black')
    plt.xlabel(etiqueta_x)
    plt.ylabel("Frecuencia")
    plt.title(titulo)
    plt.show()
```

```python
def graficar_barras(df, columna, titulo, etiqueta_x, etiqueta_y, horizontal=False, orden=None):
    conteo = df[columna].value_counts()
    
    # Reordenar si se especifica un orden particular
    if orden is not None:
        conteo = conteo.reindex(orden).dropna()
        
    plt.figure(figsize=(8, 5))
    if horizontal:
        conteo.plot(kind='barh', edgecolor='black', color='C0')
    else:
        conteo.plot(kind='bar', edgecolor='black', color='C0')
        
    plt.xlabel(etiqueta_x)
    plt.ylabel(etiqueta_y)
    plt.title(titulo)
    plt.show()
```

```python
# Hipotesis 1 - Univariada

# Histograma
plt.figure(figsize=(8, 4))

plt.hist(df_limpio[col_hp].dropna(),bins=30,edgecolor="black")

plt.title("Distribución de la potencia nominal")
plt.xlabel("Potencia (HP)")
plt.ylabel("Frecuencia")
plt.tight_layout()
plt.show()

#Boxplot
plt.figure(figsize=(8, 3))

plt.boxplot(df_limpio[col_hp].dropna(), orientation='horizontal')
plt.title("Boxplot de la potencia nominal")
plt.xlabel("Potencia (HP)")
plt.tight_layout()
plt.show()

#Estadistica Descriptiva
print("Media:", df_limpio[col_hp].mean())
print("Mediana:", df_limpio[col_hp].median())
print("Asimetría:", df_limpio[col_hp].skew())

display(df_limpio[col_hp].describe())
```

```python
# Hipótesis 2 - Bivariada
col_peso = "Equivalent Test Weight (lbs.)"
col_co2 = "CO2 (g/mi)"

# Seleccionamos los registros que tienen datos en ambas variables
datos = df_limpio[[col_peso, col_co2]].dropna()

# Diagrama de dispersión
plt.figure(figsize=(9, 5))

plt.scatter(
    datos[col_peso],
    datos[col_co2],
    alpha=0.4
)

plt.title("Relación entre el peso del vehículo y las emisiones de CO2")
plt.xlabel("Peso equivalente de prueba (lbs.)")
plt.ylabel("Emisiones de CO2 (g/mi)")
plt.tight_layout()
plt.show()

# Coeficientes de correlación
pearson = datos[col_peso].corr(datos[col_co2], method="pearson")
spearman = datos[col_peso].corr(datos[col_co2], method="spearman")

print("Cantidad de registros analizados:", len(datos))
print("Correlación de Pearson:", round(pearson, 3))
print("Correlación de Spearman:", round(spearman, 3))
```

```python
# Relaciones entre cuantitativas 

# Variables cuantitativas seleccionadas
variables_cuantitativas = [
    "RND_ADJ_FE",
    "Equivalent Test Weight (lbs.)",
    "Rated Horsepower",
    "Test Veh Displacement (L)",
    "# of Cylinders and Rotors"
]

# Seleccionamos las columnas y convertimos a valores numéricos
datos_scatter = df_limpio[variables_cuantitativas].copy()

for columna in variables_cuantitativas:
    datos_scatter[columna] = pd.to_numeric(
        datos_scatter[columna],
        errors="coerce"
    )

# Eliminamos registros que tengan nulos en alguna de las variables
datos_scatter = datos_scatter.dropna()

# Creamos el scatter matrix
pd.plotting.scatter_matrix(
    datos_scatter,
    figsize=(14, 14),
    alpha=0.4,
    diagonal="hist",
    hist_kwds={"bins": 25, "edgecolor": "black"}
)

plt.suptitle(
    "Matriz de dispersión de las variables cuantitativas",
    y=1.02,
    fontsize=15
)

plt.tight_layout()
plt.show()
```

```python
# Relacion entre cuantitativas y cualitativas: segun Sistema de Traccion 

variables_pairplot = variables_cuantitativas + [
    "Drive System Description"
]

datos_pairplot = df_limpio[variables_pairplot].copy()

# Eliminamos registros incompletos
datos_pairplot = datos_pairplot.dropna()

# Creamos el pairplot
sns.pairplot(
    datos_pairplot,
    vars=variables_cuantitativas,
    hue="Drive System Description",
    diag_kind="hist",
    plot_kws={"alpha": 0.5, "s": 25}
)

plt.suptitle(
    "Relación entre variables cuantitativas según sistema de tracción",
    y=1.02
)

plt.show()
```
